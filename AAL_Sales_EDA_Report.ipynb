## 6. Recommendation for Treating Missing or Incorrect Data

Recommended approach based on common analytics practice:

1. If only a very small number of rows contain nulls and those rows are not business-critical, dropping them is acceptable.
2. If nulls appear in important numeric columns such as `Sales` or `Unit`, filling should be considered only when there is a reliable business rule or historical basis.
3. For categorical fields such as `State`, `Group`, or `Time`, filling should be done only if a valid source or deterministic rule exists. Otherwise, keep the rows separate for review.
4. Incorrect numeric entries should be corrected when a business rule exists. In this dataset, `Sales = Unit * 2500` provides a strong validation rule.
5. Duplicates should generally be dropped only after confirming that they are accidental duplicates and not repeated legitimate transactions.

For this specific dataset, no rows need to be dropped or filled because there are no null or incorrect values in the inspected fields.

## 7. Data Preparation Before Normalization

```python
df_clean = df[['Date_parsed', 'Time', 'State', 'Group', 'Unit_numeric', 'Sales_numeric']].copy()
df_clean.columns = ['Date', 'Time', 'State', 'Group', 'Unit', 'Sales']
df_clean['Date'] = df_clean['Date'].dt.strftime('%Y-%m-%d')
df_clean.head()
```

## 8. Normalization

Normalization is the preferred wrangling method for this problem. Since `Unit` and `Sales` are numeric measures on different scales, min-max normalization is suitable because it scales both columns into the range 0 to 1 while preserving their relative positions.

Formula used:

`x_normalized = (x - x_min) / (x_max - x_min)`

```python
df_normalized = df_clean.copy()

for column in ['Unit', 'Sales']:
    min_value = df_normalized[column].min()
    max_value = df_normalized[column].max()
    df_normalized[f'{column}_Normalized'] = (df_normalized[column] - min_value) / (max_value - min_value)

df_normalized.head(10)
```

```python
df_normalized[['Unit', 'Unit_Normalized', 'Sales', 'Sales_Normalized']].describe().T
```

### Interpretation of the Normalized Result

After normalization, the smallest `Unit` and `Sales` values become 0 and the largest values become 1. This makes the numeric columns easier to compare visually and more suitable for machine learning or distance-based techniques.

## 9. GroupBy Analysis and Recommendation

The `groupby()` function is useful for chunking data into logical business segments before aggregation. It is especially appropriate here because the dataset contains repeated observations across states, customer groups, and times of day.

Typical use cases in this dataset:

- Group sales by `State` to compare regional performance.
- Group sales by `Group` to compare customer segments.
- Group by both `State` and `Group` to study segment performance inside each state.
- Group by `Date` for time-series aggregation.

Recommendation: use `groupby()` for data chunking and summarization, not merging. Since the required information already exists in one dataset, chunking is the better choice here. Merging would only be necessary if an external lookup table or another transactional dataset had to be joined.

```python
state_summary = df_clean.groupby('State').agg(
    Total_Units=('Unit', 'sum'),
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean')
).sort_values('Total_Sales', ascending=False)

group_summary = df_clean.groupby('Group').agg(
    Total_Units=('Unit', 'sum'),
    Total_Sales=('Sales', 'sum')
).sort_values('Total_Sales', ascending=False)

time_summary = df_clean.groupby('Time').agg(
    Total_Sales=('Sales', 'sum')
).sort_values('Total_Sales', ascending=False)

display(state_summary)
display(group_summary)
display(time_summary)
```

## 16. Hypothesis Testing

This section adds hypothesis testing as a final statistical layer to the project. The goal is to test whether sales differ significantly across business-relevant categories and to support decision-making with inferential statistics.


### 16.1 Objective and Test Selection

Two hypothesis-testing approaches are appropriate for this dataset:
- A one-way ANOVA test to determine whether average sales differ across the three time-of-day periods: Morning, Afternoon, and Evening.
- An independent t-test to compare average sales between two customer groups. In this notebook, Men and Women are used as a business-relevant comparison example.

A significance level of 0.05 is used for both tests.


### 16.2 Hypothesis 1: Do Mean Sales Differ by Time of Day?

Null Hypothesis (H0): The mean sales are equal across Morning, Afternoon, and Evening periods.

Alternative Hypothesis (H1): At least one time period has a different mean sales value.


```python
from scipy.stats import f_oneway, ttest_ind

morning_sales = df_clean[df_clean['Time'] == 'Morning']['Sales']
afternoon_sales = df_clean[df_clean['Time'] == 'Afternoon']['Sales']
evening_sales = df_clean[df_clean['Time'] == 'Evening']['Sales']

anova_stat, anova_p = f_oneway(morning_sales, afternoon_sales, evening_sales)

print('ANOVA test statistic:', anova_stat)
print('ANOVA p-value:', anova_p)

if anova_p < 0.05:
    print('Result: Reject the null hypothesis. Mean sales differ significantly by time of day.')
else:
    print('Result: Fail to reject the null hypothesis. Mean sales do not differ significantly by time of day.')
```

### 16.3 Hypothesis 2: Is There a Significant Difference Between Men and Women Sales?

Null Hypothesis (H0): The mean sales for Men and Women are equal.

Alternative Hypothesis (H1): The mean sales for Men and Women are different.


```python
men_sales = df_clean[df_clean['Group'] == 'Men']['Sales']
women_sales = df_clean[df_clean['Group'] == 'Women']['Sales']

ttest_stat, ttest_p = ttest_ind(men_sales, women_sales, equal_var=False)

print('T-test statistic:', ttest_stat)
print('T-test p-value:', ttest_p)

if ttest_p < 0.05:
    print('Result: Reject the null hypothesis. Men and Women sales differ significantly.')
else:
    print('Result: Fail to reject the null hypothesis. Men and Women sales do not differ significantly.')
```

### 16.4 Interpretation and Business Value

- The ANOVA test helps determine whether time-of-day differences are statistically meaningful, not just visually noticeable.
- The t-test helps compare whether two important customer segments behave differently in terms of average sales.
- If the p-value is below 0.05, the observed difference is considered statistically significant at the 5% level.
- These results can support campaign timing, segmentation strategy, and targeted promotional decisions.


### 16.5 Conclusion of Hypothesis Testing

Hypothesis testing strengthens the overall project by moving beyond descriptive summaries into inferential analysis. In this notebook, the added ANOVA and t-test indicate that the tested differences are not statistically significant at the 5% level, so the visible gaps between these selected categories should not be treated as strong evidence of different average sales behavior without further analysis.
